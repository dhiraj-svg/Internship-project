# DATA VIZUALIZATION

## BAR CHART- Top 10 Most Populated Countries

```python
top_countries=df.sort_values(by='2022',ascending=False).head(18)

# plotting a bar chart
plt.figure(figsize=(10,6))
sns.barplot(x='Country Name',y='2022',data=top_countries,palette='cool')
plt.title('Top 10 Most Populated Countries (2022)',fontsize=14)
plt.xlabel('Country')
plt.ylabel('Population')
plt.xticks(rotation=45)
plt.show()
```

## HISTOGRAM-2022 Population

```python
df['2022']=pd.to_numeric(df['2022'],errors='coerce')

df.dropna(subset=['2022'],inplace=True)

plt.figure(figsize=(8,6))
plt.hist(df['2022'],bins=20,color='lightblue',edgecolor='black')
plt.title('Distribution of 2022 Population Values',fontsize=14)
plt.xlabel('Population in 2022')
plt.ylabel('Number of Records')
plt.show()
```
